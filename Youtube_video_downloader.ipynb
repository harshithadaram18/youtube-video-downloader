```python
import yt_dlp

def get_video_info(video_url):
    options = {
        "quiet": True,
        "no_warnings": True,
        "noplaylist": True
    }

    try:
        with yt_dlp.YoutubeDL(options) as ydl:
            info = ydl.extract_info(video_url, download=False)

        print("Video Title:", info.get("title", "Unknown"))
        print("Channel:", info.get("uploader", "Unknown"))
        print("Duration:", info.get("duration", 0), "seconds")

    except Exception as e:
        print("Could not fetch video details.")
        print("Error:", str(e)[:200])
```

```python
import yt_dlp
import os

def download_video(video_url):
    options = {
        "format": "best",
        "outtmpl": "/content/%(title)s.%(ext)s",
        "noplaylist": True
    }

    try:
        with yt_dlp.YoutubeDL(options) as ydl:
            ydl.download([video_url])

        print("Download completed!")

    except Exception:
        print("Download failed.")
        print("The website may restrict access or require verification.")
```
